Load the pipeline and multilingual sentiment model

```python
from transformers import pipeline

classifier = pipeline(model="nlptown/bert-base-multilingual-uncased-sentiment")
```

```python
classifier("I have an amazing idea .")
```

# Case Study: USbi73MD

## First: Testing the method on the English data

```python
classifier(" hey Jude . this is Clara . I just saw a pretty weird thing . these this guy was bouncing his ball . and then he dropped it . and it rolled across the street . and then a car came . and quickly hit the brakes . and the car behind it crashed into it . and then a dog went kind of psycho . and a lady . dropped her bag of groceries trying to like hold back the dog . and then one of the car owners dropped his one of the car owners called 911 . and the police came")
```

Test reveals that it makes more sense to break the document down into units. For instance, CUs.

```python
classifier(
    ["hey Jude",
     "this is Clara",
     "I just saw a pretty weird thing",
     "these this guy was bouncing his ball",
     "and then he dropped it",
     "and it rolled across the street",
     "and then a car came",
     "and quickly hit the brakes",
     "and the car behind it crashed into it",
     "and then a dog went kind of psycho",
     "and a lady dropped her bag of groceries trying to like hold back the dog",
     "and then one of the car owners dropped his one of the car owners called 911",
     "and the police came"]
)
```
